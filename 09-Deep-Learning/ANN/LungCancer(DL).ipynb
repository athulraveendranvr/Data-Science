```python
#Removing unwanted columns
df.drop(['Name','Surname'],axis=1,inplace=True)
```

```python
#Spliting inputs and output
x = df.drop('Result',axis=1)
y = df['Result']
```

```python
#Output Count plotting
dd = y.value_counts()
plt.bar(dd.index,dd.values,color='r')
plt.xlabel('Results')
plt.ylabel('Count')
plt.title('Results & Counts')
```

```python
#Spliting Training and Testing data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.3,random_state = 42,stratify = y)
```

```python
#Scaling
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler.fit(x_train)
x_train = scaler.transform(x_train)
x_test = scaler.transform(x_test)
```

```python
#Importing DL module
import tensorflow as tf
```

```python
#Importing SubModule Keras
from keras.models import Sequential
from keras.layers import Dense
```

```python
#Creating object for Sequential
ann = Sequential()
#Input Layer
ann.add(Dense(units=4,activation='relu'))
#Hidden layer1
ann.add(Dense(units=10,activation='relu'))
#Hidden layer2
ann.add(Dense(units=6,activation='relu'))
#Output layer2
ann.add(Dense(units=1,activation='sigmoid'))
```

```python
#Specifying the accuracy and error calculation models and weight adjustment method
ann.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
#Fitting Training Data
ann.fit(x_train,y_train,epochs=100,batch_size=12)
```

```python
#Evaluating the performance of our model
ann.evaluate(x_test,y_test)
```

```python
#Predicting a new data
y_pred = ann.predict(scaler.transform([[42,2,4,6]]))
y_pred
if y_pred >= 0.5:
  print('Lung cancer')
else:
  print('No')
```
