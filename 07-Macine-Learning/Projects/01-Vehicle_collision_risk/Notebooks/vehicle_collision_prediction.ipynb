```python
data = pd.read_csv('Vehicle_crash_risk.csv')
df = pd.DataFrame(data)
```

```python
df['vehicle_type']=df['vehicle_type'].fillna('Unknown')
```

```python
df['weather'] = df['weather'].fillna('Unknown')
```

```python
df['road_condition'] = df['road_condition'].fillna('Unknown')
```

```python
df['time_of_day']=df['time_of_day'].fillna('Unknown')
```

```python
df['vehicle_speed_kmh']=df['vehicle_speed_kmh'].fillna(df['vehicle_speed_kmh'].median())
```

```python
df['front_vehicle_distance_m']=df['front_vehicle_distance_m'].fillna(df['front_vehicle_distance_m'].median())
```

```python
df['front_vehicle_speed_kmh']=df['front_vehicle_speed_kmh'].fillna(df['front_vehicle_speed_kmh'].median())
```

```python
df['vehicle_acceleration_mps2']=df['vehicle_acceleration_mps2'].fillna(df['vehicle_acceleration_mps2'].median())
```

```python
df['front_vehicle_acceleration_mps2']=df['front_vehicle_acceleration_mps2'].fillna(df['front_vehicle_acceleration_mps2'].median())
```

```python
df['visibility_km']=df['visibility_km'].fillna(df['visibility_km'].median())
```

```python
df['nearby_vehicle_count']=df['nearby_vehicle_count'].fillna(df['nearby_vehicle_count'].median())
```

```python
df['traffic_density_index']=df['traffic_density_index'].fillna(df['traffic_density_index'].median())
```

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
df = df.drop(['scenario_id','vehicle_id','vehicle_brand','vehicle_model'],axis=1)
```

```python
be_x = df.drop('collision_risk',axis=1)
y = df['collision_risk']
```

```python
from sklearn.preprocessing import OneHotEncoder
categorical_cols = ['vehicle_type','road_type','weather','road_condition','time_of_day']
encoder = OneHotEncoder(handle_unknown='ignore',sparse_output=False)
en_cat = encoder.fit_transform(be_x[categorical_cols])
en_cat = pd.DataFrame(en_cat,columns=encoder.get_feature_names_out(categorical_cols),index=be_x.index)
be_x.drop(columns = categorical_cols,inplace=True)
x = pd.concat([be_x,en_cat],axis=1)
```

```python
dd = y.value_counts()
plt.bar(dd.index,dd.values,color='b')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Catrgories & Counts')
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state=42,stratify=y)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler.fit(x_train)
x_train = scaler.transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,f1_score,precision_score,recall_score
```

```python
from sklearn.neighbors import KNeighborsClassifier
model_1 = KNeighborsClassifier(n_neighbors=5)
model_1.fit(x_train,y_train)
y_pred_1 = model_1.predict(x_test)
y_pred_1
```

```python
as_1 = accuracy_score(y_test,y_pred_1)
print(as_1)
```

```python
pc_1 = precision_score(y_test,y_pred_1,average='weighted')
print(pc_1)
```

```python
re_1 = recall_score(y_test,y_pred_1,average='weighted')
print(re_1)
```

```python
f1_1 = f1_score(y_test,y_pred_1,average='weighted')
print(f1_1)
```

```python
cm_1 = confusion_matrix(y_test, y_pred_1)
sns.heatmap(cm_1, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')
plt.title('Random Forest Confusion Matrix')
plt.show()
```

```python
cr_1 = classification_report(y_test,y_pred_1)
print(cr_1)
```

```python
from sklearn.naive_bayes import GaussianNB
model_2 = GaussianNB()
model_2.fit(x_train,y_train)
y_pred_2 = model_2.predict(x_test)
y_pred_2
```

```python
as_2 = accuracy_score(y_test,y_pred_2)
print(as_2)
```

```python
pc_2 = precision_score(y_test,y_pred_2,average='weighted')
print(pc_2)
```

```python
re_2 = recall_score(y_test,y_pred_2,average='weighted')
print(re_1)
```

```python
f1_2 = f1_score(y_test,y_pred_2,average='weighted')
print(f1_2)
```

```python
cm_2 = confusion_matrix(y_test, y_pred_2)
sns.heatmap(cm_2, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')
plt.title('Naive Bays Confusion Matrix')
plt.show()
```

```python
cr_2 = classification_report(y_test,y_pred_2)
print(cr_2)
```

```python
from sklearn.tree import DecisionTreeClassifier
model_3 = DecisionTreeClassifier(criterion='entropy')
model_3.fit(x_train,y_train)
y_pred_3 = model_3.predict(x_test)
y_pred_3
```
