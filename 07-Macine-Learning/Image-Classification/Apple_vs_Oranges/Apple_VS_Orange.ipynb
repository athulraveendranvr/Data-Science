```python
input = []
output = []
categories = ['apple','orange']
datadir = '/content/drive/MyDrive/fruit-dataset'
```

```python
for i in categories:
  print('Loading categories')
  path = os.path.join(datadir,i)
  for img in os.listdir(path):
    img_read = imread(os.path.join(path,img))
    img_res = resize(img_read,(150,150,3))
    input.append(img_res.flatten())
    output.append(categories.index(i))
print('Loading completed')
```

```python
input = np.array(input)
output = np.array(output)
```

```python
df = pd.DataFrame(input)
df['target'] = output
df
```

```python
x = df.drop('target',axis=1).values
y = df['target']
```

```python
dd = df['target'].value_counts()
plt.bar(dd.index,dd.values,color = 'b')
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state = 42)
```

```python
from sklearn.svm import SVC
model = SVC()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
img1 = '/content/drive/MyDrive/fruit-dataset/apple1.jpg'
img1_read = imread(img1)
img1_res = resize(img1_read,(150,150,3)).reshape(1,-1)
y_pred1 = model.predict(img1_res)
y_pred1
if y_pred1 == 0:
  print('apple')
else:
  print('Orange')
```
