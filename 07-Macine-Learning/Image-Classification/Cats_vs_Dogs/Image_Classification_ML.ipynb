```python
#Create an empty list to collect the inputs(Pixel values)of each image
input = []
#Create an empty list to collect the corresponding class labels of each image
output = []
#List the class labels in its original order
categories = ['Dog','Cat']
#Taking the datapath of the main folder for joining
datadir = '/content/drive/MyDrive/Data'
```

```python
for i in categories: #i = Dog(first)
  print('Loading categories...')
  path  = os.path.join(datadir,i)
  for img in os.listdir(path):
    img_read=imread(os.path.join(path,img))
    img_res=resize(img_read,(150,150,3))
    input.append(img_res.flatten())
    output.append(categories.index(i))
print('Loading completed')
```

```python
input = np.array(input)
output = np.array(output)
```

```python
df = pd.DataFrame(input)
df['target']=output
df
```

```python
x = df.drop(['target'],axis=1).values
y = df['target'].values
```

```python
dd = df['target'].value_counts()
plt.bar(dd.index,dd.values,color='b')
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
#Model creation
#SVM (support vector machine) ==> Image classification
from sklearn.svm import SVC
model = SVC()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
